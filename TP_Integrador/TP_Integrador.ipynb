## Primera prueba del Modelo 1

Utilizamos el modelo `distilbert/distilbert-base-uncased-finetuned-sst-2-english`
mediante `pipeline` de Transformers.

La frase utilizada fue:

> "This movie was absolutely amazing!"

El modelo devuelve una etiqueta de sentimiento (`POSITIVE` o `NEGATIVE`)
junto con un valor de confianza (`score`).

```python
frases = [
    "This movie was absolutely amazing!",
    "I hated every minute of this movie.",
    "La película fue excelente y me encantó.",
    "La película fue aburrida y demasiado larga.",
    "Yeah, great movie... I almost fell asleep.",
    "This movie was sick!",
    "Meh.",
    "The acting was good, but the story was terrible."
]
```

```python
resultados_modelo_1 = modelo_1(frases)

resultados_modelo_1
```
