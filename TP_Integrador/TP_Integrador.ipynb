### Error con InferenceClient: 402 Payment Required

Al intentar utilizar `InferenceClient` con el modelo `distilbert/distilbert-base-uncased-finetuned-sst-2-english`, la solicitud devolvió un error HTTP 402 (`Payment Required`).

El token de Hugging Face fue validado correctamente, por lo que el problema no estaba relacionado con la autenticación. El mensaje de la API indicó que la cuenta no tenía créditos disponibles para utilizar Inference Providers.

Como alternativa, se continuó utilizando `pipeline`, que permite ejecutar el modelo localmente en el entorno de trabajo sin realizar una llamada remota al proveedor de inferencia.
